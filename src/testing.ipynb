# Test Generators
Testing whether I can have a list that updates inside an active generator, and whether generators make a good new hypothesis generator.

```python
from typing import Generator

def mutate_data_in_generator(l:list[int])->Generator[int,None,None]:
    for i in l: yield i

l = [l for l in range(0,10)]
print(l)
print([e for e in mutate_data_in_generator(l)]) # Simple generator usage
n = []
for i,e in enumerate(mutate_data_in_generator(l)):
    n.append(e)
    l[i] = 3*e # make changes after generator passes
    if i != len(l)-1: l[i+1] += 10 # make changes befor generator reaches
print(n) # Print generator outputs
print(l) # Print current list state

```

Now test sending data into generator

```python
def my_generator():
    x = yield 'first'   # x receives whatever is sent in
    y = yield 'second'  # y receives whatever is sent in
    yield x + y

gen = my_generator()

print(next(gen))        # 'first'  - starts the generator, use next() not send()
print(gen.send(10))     # 'second' - sends 10, x = 10
print(gen.send(20))     # 30       - sends 20, y = 20
```
